### Import Libraries

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import os

import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator,load_img, img_to_array
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Conv2D, MaxPooling2D, Flatten, Dense, Dropout

import warnings
warnings.filterwarnings("ignore")
```

### Load datasets

```python
train_dir = r"C:\Users\91918\Downloads\DL + AI Exam Paper\DL + AI Exam Paper\Dataset\CNN\Shoe Classification\train"
```

```python
test_dir = r"C:\Users\91918\Downloads\DL + AI Exam Paper\DL + AI Exam Paper\Dataset\CNN\Shoe Classification\test"
```

```python
### Data preprocessing
```

```python
train_datagen = ImageDataGenerator(rescale=1./255,validation_split=0.2)
```

```python
train_data = train_datagen.flow_from_directory(train_dir,target_size=(224,224),batch_size=32,class_mode='categorical',subset='training')
```

```python
val_data = train_datagen.flow_from_directory(
    train_dir,
    target_size=(224,224),
    batch_size=32,
    class_mode='categorical',
    subset='validation'
)
```

### Build cnn model

```python
model = Sequential()
```

```python
model.add(Conv2D(32,(3,3),activation='relu',input_shape=(224,224,3)))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Conv2D(64,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Conv2D(128,(3,3),activation='relu'))
model.add(MaxPooling2D(2,2))
```

```python
model.add(Flatten())
```

```python
model.add(Dense(128,activation='relu'))
model.add(Dropout(0.5))
```

```python
model.add(Dense(3,activation='softmax'))
```

```python
### Compile model
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
### Train model
```

```python
history = model.fit(
    train_data,
    validation_data=val_data,
    epochs=20
)
```

```python
### Predict test data
```

```python
images = []
file_names = []

for file in os.listdir(test_dir):
    img_path = os.path.join(test_dir, file)
    
    try:
        img = load_img(img_path, target_size=(224,224))
        img = img_to_array(img)/255.0
        
        images.append(img)
        file_names.append(file)
        
    except:
        print("Skipped:", file)

images = np.array(images)

predictions = model.predict(images)
predicted_classes = np.argmax(predictions, axis=1)
```

```python
print("Accuracy:", history.history['accuracy'][-1])
```
